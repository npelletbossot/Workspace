```python
import numpy as np

def reconstitute_trajectories(
    t_matrix: np.ndarray,
    x_matrix: np.ndarray,
    tmax: int,
    nt: int,
) -> np.ndarray:
    """
    Pour chaque trajectoire i et bin entier t ∈ [0, tmax),
    retourne la valeur x du dernier événement survenu avant t+1.

    t_matrix / x_matrix : object array (lignes de longueur variable)
                          OU 2-D float array padded avec NaN.
    """
    trajectories = np.zeros((nt, tmax), dtype=int)
    bins = np.arange(1, tmax + 1, dtype=float)   # bords droits des bins

    for i in range(nt):
        t_row = np.asarray(t_matrix[i], dtype=float).ravel()
        x_row = np.asarray(x_matrix[i], dtype=float).ravel()

        # --- Masque : élimine NaN + événements hors [0, tmax) ---------------
        valid = np.isfinite(t_row) & (t_row < tmax)
        t_v = t_row[valid]
        x_v = x_row[valid].astype(int)

        if t_v.size == 0:
            continue                              # trajectoire vide → reste 0

        # --- Pour chaque bin t, dernier événement avant t+1 -----------------
        # searchsorted 'right' : nb d'éléments <= valeur  → -1 pour index 0-based
        idx = np.searchsorted(t_v, bins, side="right") - 1

        hit = idx >= 0                            # bins qui ont au moins 1 événement
        trajectories[i, hit] = x_v[idx[hit]]

    return trajectories
```
